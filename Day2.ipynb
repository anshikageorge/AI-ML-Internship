```python
from sklearn.preprocessing import MinMaxScaler, StandardScaler
```

```python
#Load Dataset
file_path='/content/sample_data/preprocessing_practice_dataset.csv'
prepros_data=pd.read_csv(file_path)
```

```python
numerical_columns=['Age','Experience_Years','Monthly_Salary','Performance_Rating']
```

```python
prepros_data.dropna(subset=numerical_columns,inplace=True)
```

```python
minmax_scaler=MinMaxScaler()
standard_scaler=StandardScaler()
```

```python
prepros_data['Age_MinMax'] = minmax_scaler.fit_transform(prepros_data[['Age']])
prepros_data['Experience_Years_MinMax'] = minmax_scaler.fit_transform(prepros_data[['Experience_Years']])
prepros_data['Monthly_Salary_MinMax'] = minmax_scaler.fit_transform(prepros_data[['Monthly_Salary']])
prepros_data['Performance_Rating_MinMax'] = minmax_scaler.fit_transform(prepros_data[['Performance_Rating']])

prepros_data['Age_Standard'] = standard_scaler.fit_transform(prepros_data[['Age']])
prepros_data['Experience_Years_Standard'] = standard_scaler.fit_transform(prepros_data[['Experience_Years']])
prepros_data['Monthly_Salary_Standard'] = standard_scaler.fit_transform(prepros_data[['Monthly_Salary']])
prepros_data['Performance_Rating_Standard'] = standard_scaler.fit_transform(prepros_data[['Performance_Rating']])
```
